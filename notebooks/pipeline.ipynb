Pada sel di atas, skrip **`pipeline.py`** dieksekusi secara otomatis menggunakan **BeamDagRunner**. Komponen TFX yang dijalankan secara berurutan meliputi:

- **ExampleGen:** Memuat dataset Adult Census Income dari file `data/adult.csv` dan membaginya menjadi dua rasio split (80% `train` dan 20% `eval`).
- **StatisticsGen, SchemaGen, dan ExampleValidator:** Mengonsolidasi statistik deskriptif data, membuat skema tipe data, serta mengevaluasi potensi anomali pada dataset.
- **Transform:** Menerapkan proses feature engineering berbasis modul `modules/transform.py` (standarisasi Z-score untuk fitur numerik dan vocabulary indexing untuk fitur kategorikal).
- **Tuner:** Menjalankan pencarian hyperparameter optimal secara terotomatisasi via KerasTuner (`modules/tuner.py`).
- **Trainer:** Melatih arsitektur Deep Neural Network menggunakan kombinasi hyperparameter terbaik (`modules/trainer.py`).
- **Resolver:** Mengambil model blessed terakhir sebagai pembanding (baseline model).
- **Evaluator:** Menilai performa model terhadap ambang batas akurasi minimal 60% dan memastikan nilainya lebih baik dibanding model baseline.
- **Pusher:** Mengekspor file SavedModel yang siap deploy ke direktori `serving_model/adult-income-model` apabila model dinyatakan blessed.
